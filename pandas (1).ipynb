```python
plt.figure(figsize=(10, 6))
df.set_index('userid')['friend_count'].head(10).plot(kind='bar')
plt.xlabel("User ID")
plt.ylabel("Friend Count")
plt.title('Friend Count of the top 10 Users')
plt.show()
```

```python
plt.figure(figsize=(8, 6))
plt.hist(df['likes_received'].dropna(), bins=20, edgecolor='black')
plt.xlabel("Likes Received")
plt.ylabel("Frequency")
plt.title("Distribution of Likes Received")
plt.show()
```

```python
sns.pairplot(
    data=df,
    x_vars=['age'],
    y_vars=['friend_count', 'likes', 'likes_received'],
    kind='reg',
)
plt.show()
```

```python
sns.pairplot(
    hue="gender",
    data=df,
    x_vars=["age"],
    y_vars=["friend_count", "likes", "likes_received"],
    kind="reg",
)
plt.show()
```
